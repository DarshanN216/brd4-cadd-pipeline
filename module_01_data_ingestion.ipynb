```python
import pandas as pd
from chembl_webresource_client.new_client import new_client
```

```python
BASE_DIR= Path(os.getcwd())
RAW_DIR=BASE_DIR/"brd4_pipeline"/"data"/"raw"
FILTERED_DIR=BASE_DIR/"brd4_pipeline"/"data"/"filtered"
PROCESSED_DIR=BASE_DIR/"brd4_pipeline"/"data"/"processed"

RAW_DIR.mkdir(parents=True, exist_ok=True)
FILTERED_DIR.mkdir(parents=True, exist_ok=True)
PROCESSED_DIR.mkdir(parents=True, exist_ok=True)

TARGET_ID='CHEMBL4523'

activity=new_client.activity

print(f"[INFO] Querying ChEMBL for : {TARGET_ID}")
print("[INFO] This may take 30-90 seconds. Do not interrupt.")

results=activity.filter(
    target_chembl_id=TARGET_ID,
    standard_type="IC50"
).only(
    'molecule_chembl_id',
    'canonical_smiles',
    'standard_value',
    'standard_units',
    'assay_type',
    'pchembl_value',
    'data_validity_comment',
    'standard_type')

print(f"[INFO] Downloading records...")
result_list=list(results)

df_raw=pd.DataFrame.from_records(result_list)

print(f"[SUCCESS] Retrived {len(df_raw):,} raw records")
print(f"Columns: {df_raw.columns.tolist()}")
print(df_raw.head(3))
raw_file=RAW_DIR/"brd4_IC50_raw.csv"
df_raw.to_csv(raw_file, index=False)
print(f"\n[SAVED] : {raw_file}")
print(f"Shape : {df_raw.shape[0]:,} rows x {df_raw.shape[1]} columns")
```
